**a. Read the dataset description. Who produced tha data? When, how, and for what purpose?**

**Answer:** The data was made by Reimo Palm from the University of Tartu. It comes from Moodle of the course “Theoretical Computer Science” and covers spring semesters 2019–2026. Event logs and grades were downloaded from Moodle, and the course calendar was made from ÕIS and official emails. Then a Python script hid real names: every student got a random code (example Sxxxxxx) , and all teachers got one code T. Moodle saves these logs to run the course, not for research. The dataset was prepared for a student project to study how Moodle activity changed over the years and how it is connected to grades.

**b. Name a plausible stakeholder, their goals, and the concrete decision the analysis would support. Who benefits, who could be harmed?**

**Answer:** The stakeholders are the course teachers. They want fewer students to fail the course. Our analysis can help them see, before the midterm, which students use Moodle in a similar way to students who failed in earlier years, so they can offer help in time. The students who get help and the teachers would benefit. It could harm students that the model wrongly marks as weak, because teachers might treat them differently. It is also unfair to students who don’t use Moodle much but study well in other ways, for example by going to classes or using paper materials. There is also a privacy question: students don’t expect their Moodle actions to be analysed like this.

**c. State the objective in business terms (decide strategy, predict demand or delay, recognize priorities, connect factors to outcomes, detect risks, etc), and how success would be judged (cost saved, error tolerated, etc).**

For example:
* _"Predict delay for bus lines so that passengers can choose an earlier connection."_
* _"Recognize priorities among customer complaints so that support staff answer urgent ones first."_

NB! For this step, remember to name the objective without naming the method. For example, don't state “predict X with a random forest” as the objective.

**Answer:** Our goal is to find students who are at risk of failing the course early, one week before the midterm, so teachers have time to help them. We also want to understand how the use of Moodle changed over the years, so teachers can see which materials and formats students really use. Success means we find most of the students who later fail, for example at least 7 out of 10, without marking too many students who would pass anyway, for example at least 1 out of 3 marked students should really fail. 

Missing a weak student is worse than sending an extra email to a strong one, because an extra email costs almost nothing. It should also work on new years, not only on old ones. For the second goal, success means we can clearly show whether and when activity patterns changed and which materials students use most.

**d. Now translate the objective to a data science problem.**

Is it classification, regression, clustering, description? What would be predicted or explained? At what point in time would the prediction or other outcome be used?

**Answer:** The main task is classification: predict from a student’s Moodle activity whether they will pass the course or not. Pass means a grade from A to E. Fail means F or not showing up for the exam. 

The prediction is made about one week before the first midterm, so we only use data that exists at that point: the logs from the start of the semester until one week before the midterm. We train the model on older years (2019, 2021–2024) and test it on newer years (2025–2026) to see if it works for future students. Another part of the project is descriptive analysis: we compare student and teacher activity across years and look at whether changes in activity coincide with COVID-19, online teaching and the rise of ChatGPT.

The second step of CRISP-DM is data understanding. The goal is to collect, describe, explore, and verify the quality of the data to ensure it’s suitable for the analytical or modeling tasks ahead.

**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
import pandas as pd

years = [2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026]

events_list = []
grades_list = []
calendar_list = []

for year in years:
    events_list.append(pd.read_parquet("data/" + str(year) + "/events.parquet"))
    grades_list.append(pd.read_csv("data/" + str(year) + "/grades.csv"))
    calendar_list.append(pd.read_csv("data/" + str(year) + "/calendar.csv"))

events = pd.concat(events_list, ignore_index=True)
grades = pd.concat(grades_list, ignore_index=True)
calendar = pd.concat(calendar_list, ignore_index=True)
course_runs = pd.read_csv("data/course_runs.csv")

print("events:", events.shape)
print("grades:", grades.shape)
print("calendar:", calendar.shape)
print("course_runs:", course_runs.shape)
```

**Answer:**

The dataset has 4 tables.

events: 972,372 rows and 9 columns. One row is one action in Moodle by a student or a teacher: who did what and when.

grades: 945 rows and 41 columns. One row is one student in one year: homework, midterm and exam scores, total score and final grade. 843 of them have grades; the other 102 rows are from 2020, when no grades are available.

calendar: 750 rows and 5 columns. One row is one course activity: a lecture, lab, homework deadline, midterm or exam.

course_runs: 8 rows and 9 columns. One row is one year of the course.

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import matplotlib.pyplot as plt
import seaborn as sns

events["year"] = events["course_run_id"].str[-4:].astype(int)
grades["year"] = grades["course_run_id"].str[-4:].astype(int)
students = events[events["actor_id"] != "T"].copy()

# Plot 1: student activity by week
start = course_runs.set_index("course_run_id")["course_start_date"]
students["date"] = students["event_time"].dt.tz_localize(None).dt.normalize()
students["week"] = (students["date"] - pd.to_datetime(students["course_run_id"].map(start))).dt.days // 7 + 1
weekly = students.groupby(["week", "year"]).size().unstack() / grades.groupby("year").size()
weekly.plot(figsize=(10, 5), title="Student events per week (per student)", xlabel="week", ylabel="events per student")
plt.show()

# Plot 2: final grades
graded = grades[grades["year"] != 2020].copy()
graded["course_letter_grade"] = graded["course_letter_grade"].fillna("no grade")
sns.countplot(data=graded, x="course_letter_grade", order=["A", "B", "C", "D", "E", "F", "no grade"])
plt.title("Final grades (2019, 2021-2026)")
plt.show()

# Plot 3: active days before the midterm
midterm = calendar[calendar["activity_name"] == "midterm_attempt_1"].set_index("course_run_id")["start_time"]
cutoff = pd.to_datetime(students["course_run_id"].map(midterm), utc=True) - pd.Timedelta(days=7)
before = students[students["event_time"] < cutoff]
active_days = before.groupby(["year", "actor_id"])["date"].nunique().reset_index(name="active_days")
data = graded.merge(active_days, on=["year", "actor_id"], how="left").fillna({"active_days": 0})
data["result"] = "pass"
data.loc[data["course_letter_grade"].isin(["F", "no grade"]), "result"] = "fail"
sns.boxplot(data=data, x="result", y="active_days")
plt.title("Active days in Moodle until one week before the midterm")
plt.show()
print(data.groupby("result")["active_days"].median())
data.to_csv("active_days_hw7.csv", index=False)
```

**Answer:** 
Student activity has clear peaks in the midterm week (around week 9–10) and the exam week (around week 16). It was highest in 2020–2024 and lower in 2019 and 2025–2026. About 15% of students fail or don’t show up, so the classes are imbalanced. Students who failed were active on fewer days until one week before the midterm (median 13) than those who passed (median 23), so early activity looks useful for prediction, but the groups overlap.

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
# Columns with homework, midterm and exam scores (English and Estonian versions)
score_cols = grades.columns[2:38]

# Missing values in the event log
print("missing values in events:")
print(events.isna().sum())

# Impossible values and duplicated students
print("negative scores:", (grades[score_cols] < 0).sum().sum())
print("duplicate students in a year:", grades.duplicated(["course_run_id", "actor_id"]).sum())

# Duplicate events: identical rows except for event_id
cols = ["course_run_id", "event_time", "actor_id", "affected_actor_id", "event_context", "component", "event_name", "origin"]
dup = events.duplicated(subset=cols)
print("duplicate event rows:", dup.sum())

# Duplicates by year: check if they come from minute-level timestamps in 2019-2023
print(events[dup].groupby("year").size())

# Tasks where a student has both an English and an Estonian score
both = 0
for i in range(0, 36, 2):
    both = both + (grades[score_cols[i]].notna() & grades[score_cols[i + 1]].notna()).sum()
print("same task has both eng and est score:", both)

# Number of events per student per year: look for outliers
per_student = events[events["actor_id"] != "T"].groupby(["year", "actor_id"]).size()
print(per_student.describe())
```

**Answer:** 
There are no negative scores and no student appears twice in the same year. 2020 has no grades, so we use it only for describing activity, not for prediction. affected_actor_id is empty in 622,746 events, which is normal because most actions don’t affect another user. 145,936 event rows (15%) look like duplicates. About 88% of them are in 2019–2023, when time is saved only to the minute, so many of them are probably different actions within the same minute. The rest (about 18,000 in 2024–2026) may be repeated clicks or double logging. We won’t delete them, but we will measure activity in active days instead of raw rows. In 584 cases a student has both an English and an Estonian score for the same task, so when joining the columns we will take the higher one for now and check some of these cases more closely. 65 students have no grade (52 of them have some scores, 13 have none); these are no-shows, and we count them as fail. Some students are outliers: one has 10,205 events while the median is 641, which is another reason to use active days instead of event counts. Moodle tools also changed between years, so we will compare general features, not raw counts per component.
